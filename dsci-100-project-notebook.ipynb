#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
### EDIT THIS CODE CELL ###
# mean, min and max of every numeric column in one table
numeric_summary <- tournament_tidy |>
    # keep only the numeric columns
    select(where(is.numeric)) |>
    # stack them into a longer form using pivot_longer, then use summarize() to calculate the summary statistics listed above (making sure to drop 
    # missing values
    pivot_longer(cols = everything(), names_to = "variable", values_to = "value") |>
    group_by(variable) |>
    summarize(mean = mean(value, na.rm = TRUE),
    min = min(value, na.rm = TRUE),
    max = max(value, na.rm = TRUE))
numeric_summary
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
### EDIT THIS CODE CELL ###
# count missing values in each column of the raw data, for the columns used in our analysis (done on the raw data, before drop_na(), so we can see how 
# many rows the tidying step removed)
missing_table <- tournament_data |>
    # select columns relevant to our research question
    select(bid, conf_wins_pct, conf_rank, total_wins_pct) |>
    # use summarize to count the rows with missing values
    summarize(across(everything(), ~ sum(is.na(.x)))) |>
    # use pivot_longer to tidy the results
    pivot_longer(cols = everything(), names_to = "column", values_to = "number_missing")
missing_table
```

Visualization 1: This bar chart shows, for each conference rank, the share of teams that got an automatic bid versus an at-large bid. Conference rank is one of our predictors, so this shows how strongly it separates the two bid types (at-large and automatic). If the bars change drastically from rank to rank, rank likely carries a lot of predictive information. However, if they look similar across ranks, it will help the model less.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1###
# Visualization 1: share of each bid type at every conference rank
rank_plot <- ggplot(tournament_tidy, aes(x = factor(conf_rank), fill = bid)) +
    geom_bar(position = "fill") +
    scale_y_continuous(labels = scales::percent) +
    labs(title = "Bid Type by Place in Conference", x = "Place in Conference (1 = first)", y = "Share of Teams", fill = "Type of Bid") +
    theme(text = element_text(size = 14))
rank_plot
```

Visualization 2: This scatterplot compares conference win percentage with total win percentage, with each bar coloured by bid type. Both conference win percentage and total win percentage are predictors in our model, so we can see whether the two bid types form separate clusters. Clear separation would suggest a nearest-neighbours model can work well, but a lot of overlap would suggest that the two bid types look alike on these measures, and a nearest-neighbours model may not work very well. 

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 2###
# Visualization 2: Conference win percentage vs total win percentage, coloured by bid type
options(repr.plot.width = 14, repr.plot.height = 12)
scatter_plot <- ggplot(tournament_tidy, aes(x = conf_wins_pct, y = total_wins_pct, color = bid)) +
    geom_point(alpha = 0.5, size = 2) +
    labs(title = "Conference vs Overall Win Percentage by Bid Type", x = "Conference Win Percentage (%)", y = "Total Win Percentage (%)", color = "Type of Bid") +
    theme(text = element_text(size = 14))
scatter_plot
```
